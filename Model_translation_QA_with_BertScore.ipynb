### BERTScore is an evaluation metric that is used to analyze quality of translations
### It uses deep contextual understanding to assess similarity (goes beyound word similarities to the words meansing (Semantics).
### It compare contextual embeddings of words , making it more effective in capturing the meanings and nuances in languages.
### Uses the BERT Model


###Reference material https://spotintelligence.com/2024/08/20/bertscore/


```python
pip install bert-score -q
```

```python
from bert_score import score
```

```python
candidateText=["The cat sat on the mat"]
referenceText=["A cat is sitting on the mat"]
```

```python
P, R, F1 = score(
    candidateText, referenceText,
    lang="en",                    # or model_type="roberta-large"
    model_type="roberta-large",   # strong default for English
    rescale_with_baseline=True,   # recommended for stability
    batch_size=32                 # tune for speed/memory
)
```

```python

print("P:", P.tolist())
print("R:", R.tolist())
print("F1:", F1.tolist())
print("F1 mean:", sum(F1.tolist())/len(F1))
```
