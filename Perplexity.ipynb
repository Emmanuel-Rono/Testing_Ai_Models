## Perplexity

## Perplexity is  a measure of how well the llm predicts the next word.
### Lower Perplexity  == high Confidence in th emodel
## Higher Perplexity == Low Confidence

```python
import torch
from transformers import GPT2LMHeadModel, GPT2TokenizerFast

```

### 1. Load pre-trained model and tokenizer


```python
model_id = 'gpt2'
model = GPT2LMHeadModel.from_pretrained(model_id)
tokenizer = GPT2TokenizerFast.from_pretrained(model_id)

```

### 2. Prepare the text


```python
text = "The quick brown fox jumps over the lazy dog."
encodings = tokenizer(text, return_tensors='pt')

```

### 3. Get input IDs and compute log-probabilities


```python
input_ids = encodings.input_ids
with torch.no_grad():
    outputs = model(input_ids, labels=input_ids)
```

### 4. Calculate negative log-likelihood (cross-entropy)

```python


# The `outputs.loss` is the average negative log-likelihood.
# The model's forward pass with `labels` automatically calculates this for us.
neg_log_likelihood = outputs.loss

```

### 5. Calculate perplexity


```python
perplexity = torch.exp(neg_log_likelihood)

print(f"The perplexity of the sentence is: {perplexity.item():.2f}")
```
