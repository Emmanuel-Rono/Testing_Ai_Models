```python
from textblob import TextBlob
from detoxify import Detoxify


```

```python
# Raw user input
raw_text = "I hat this stupd thing. Its the wors!"


# Step 1: Clean and correct spelling
blob = TextBlob(raw_text)
cleaned_text = str(blob.correct())  # "I hate this stupid thing. It's the worst!"
print(cleaned_text)

```

```python
# Step 2: Analyze toxicity
toxicity_scores = Detoxify('original').predict(cleaned_text)

print(toxicity_scores)
```

## More Understanding of Fainess and Biasness.

####Fairness is the goal , biasness is the problem
####Fairness- Equitable treatment of all individuals and groups ensure the model decisions and output are not discriminatory.
Since AI models learn from historical data that may reflect real-world, societal prejudice (like racism or sexism), they can perpetuate or even amplify that unfairness.

Bias (The Problem): A systematic error in the model's outputs that disadvantages one group (e.g., a hiring model is less accurate at identifying qualified female candidates because the training data primarily featured male employees).

Fairness (The Solution): Implementing technical and ethical safeguards to ensure the model's performance and impact are comparable across different groups (e.g., adjusting the model so that the acceptance rate or error rate is similar for all genders).
